```python
# ── Stirling V=10 Newton basin map ──────────────────────────────────────────

def stirling10(z):
    return (15*z**9 - 45*z**8 - 270*z**7 + 182*z**6 + 1687*z**5
            + 1395*z**4 - 1576*z**3 - 2684*z**2 - 1008*z) / 7680

def stirling10_deriv(z):
    return (135*z**8 - 360*z**7 - 1890*z**6 + 1092*z**5 + 8435*z**4
            + 5580*z**3 - 4728*z**2 - 5368*z - 1008) / 7680

def newton_basin(f, df, roots, resolution=300, x_range=(-2.5, 2.5), y_range=(-2.5, 2.5),
                 maxiter=40, tol=1e-6):
    xs = np.linspace(*x_range, resolution)
    ys = np.linspace(*y_range, resolution)
    basin = np.zeros((resolution, resolution), dtype=int)
    for i, y in enumerate(ys):
        for j, x in enumerate(xs):
            z = complex(x, y)
            for _ in range(maxiter):
                dz = df(z)
                if abs(dz) < 1e-14: break
                z -= f(z) / dz
                if abs(f(z)) < tol:
                    # find closest root
                    dists = [abs(z - r) for r in roots]
                    basin[i,j] = int(np.argmin(dists)) + 1
                    break
    return basin

# Find roots of Stirling10 numerically
from numpy.polynomial import polynomial as P
roots_s10_approx = np.roots([15, -45, -270, 182, 1687, 1395, -1576, -2684, -1008, 0]) / 7680
# The polynomial has a factor z, so one root is 0
# Use: S10(z)/z and find 8 roots of degree-8 quotient
# Coefficient array for S10(z)/z (drop constant and last coeff):
coeff_s10_over_z = np.array([15, -45, -270, 182, 1687, 1395, -1576, -2684, -1008]) / 7680
roots_8 = np.roots(coeff_s10_over_z)
roots_all = list(roots_8) + [complex(0, 0)]
print(f"Stirling V=10 roots (8 + zero root):")
for i, r in enumerate(roots_all[:8]):
    print(f"  root {i+1}: {r:.4f}")

print("\nComputing Newton basin map (300×300, Stirling V=10)...")
basin_map = newton_basin(stirling10, stirling10_deriv, roots_8, resolution=300)
print("Done.")

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Stirling basin map
cmap8 = plt.cm.get_cmap('tab10', 9)
im = axes[0].imshow(basin_map, extent=[-2.5,2.5,-2.5,2.5], origin='lower',
                    cmap=cmap8, interpolation='nearest', vmin=0, vmax=8)
axes[0].set_title('Stirling V=10 Newton Basins\n(8 octonion output routes $e_0\\ldots e_7$)')
axes[0].set_xlabel('Re(z)'); axes[0].set_ylabel('Im(z)')
plt.colorbar(im, ax=axes[0], label='basin (octonion dim)')

# Lorenz attractor
sigma_l, rho_l, beta_l = 10.0, 28.0, 8/3
dt = 0.005
n_steps = 8000
x_l, y_l, z_l = 0.1, 0.0, 0.0
traj = np.zeros((n_steps, 3))
for i in range(n_steps):
    dx = sigma_l*(y_l - x_l); dy = x_l*(rho_l - z_l) - y_l; dz = x_l*y_l - beta_l*z_l
    x_l += dx*dt; y_l += dy*dt; z_l += dz*dt
    traj[i] = [x_l, y_l, z_l]

ax2 = axes[1]
sc = ax2.scatter(traj[:,0], traj[:,2], c=np.arange(n_steps), cmap='plasma',
                 s=0.3, alpha=0.6)
ax2.set_title('Lorenz Butterfly\n(trajectory within each Stirling basin)')
ax2.set_xlabel('x'); ax2.set_ylabel('z')

plt.tight_layout(); plt.show()
print(f"\nStirling basin coverage: {(basin_map>0).mean()*100:.1f}% of plane falls into an octonion basin")
```
